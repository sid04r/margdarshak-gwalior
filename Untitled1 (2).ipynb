```python
# ==========================================
# BLOCK 2: TOPOLOGICAL TRAVEL-TIME MATRIX (MINUTES)
# ==========================================

# Realistic transit times in minutes accounting for Gwalior's cliff topography,
# one-way bazaar lanes, and railway corridor choke points.
TRANSIT_TIME_MINS = {
    ("gwalior_junction", "man_mandir_palace"): 20,  # Via Urvai Gate ascent
    ("gwalior_junction", "gujari_mahal"): 15,       # Eastern base approach
    ("gwalior_junction", "tomb_of_tansen"): 15,     # North toward Hazira
    ("gwalior_junction", "jai_vilas_palace"): 18,   # South into Lashkar
    ("gwalior_junction", "sarod_ghar"): 16,        # Jiwaji Ganj residential zone
    ("gwalior_junction", "maharaj_bada"): 20,       # Core Lashkar commercial plaza
    ("gwalior_junction", "bateshwar_temples"): 75,  # 35km North via NH-44 into Morena

    # Fort Plateau internal movements (level ridge walks/hops)
    ("man_mandir_palace", "sas_bahu_temple"): 8,
    ("man_mandir_palace", "teli_ka_mandir"): 10,
    ("sas_bahu_temple", "teli_ka_mandir"): 8,
    ("teli_ka_mandir", "gopachal_parvat"): 12,      # Descending toward Urvai slope

    # Fort Plateau to City Points (Requires descending through Urvai Gate)
    ("man_mandir_palace", "jai_vilas_palace"): 25,
    ("man_mandir_palace", "maharaj_bada"): 25,
    ("man_mandir_palace", "tomb_of_tansen"): 20,
    ("man_mandir_palace", "gujari_mahal"): 25,      # Driving down Urvai & around the rock!

    # Lashkar Cluster movements
    ("jai_vilas_palace", "maharaj_bada"): 10,
    ("jai_vilas_palace", "sarod_ghar"): 10,
    ("sarod_ghar", "maharaj_bada"): 8,

    # Hazira to Lashkar (Cross-city traversal)
    ("tomb_of_tansen", "maharaj_bada"): 25,
    ("tomb_of_tansen", "jai_vilas_palace"): 25,
    ("tomb_of_tansen", "sarod_ghar"): 22,
    ("tomb_of_tansen", "gujari_mahal"): 10,

    # Peripheral Morena hops
    ("man_mandir_palace", "bateshwar_temples"): 85,
    ("maharaj_bada", "bateshwar_temples"): 90,
}

def get_transit_time(origin_id: str, dest_id: str) -> int:
    """
    Returns realistic transit time in minutes between two nodes.
    Uses symmetric lookup or triangular fallback estimation if direct key is missing.
    """
    if origin_id == dest_id:
        return 0

    # Check direct or reverse pair
    if (origin_id, dest_id) in TRANSIT_TIME_MINS:
        return TRANSIT_TIME_MINS[(origin_id, dest_id)]
    if (dest_id, origin_id) in TRANSIT_TIME_MINS:
        return TRANSIT_TIME_MINS[(dest_id, origin_id)]

    # Fallback heuristic: via Gwalior Junction hub
    leg1 = TRANSIT_TIME_MINS.get((origin_id, "gwalior_junction")) or TRANSIT_TIME_MINS.get(("gwalior_junction", origin_id), 20)
    leg2 = TRANSIT_TIME_MINS.get(("gwalior_junction", dest_id)) or TRANSIT_TIME_MINS.get((dest_id, "gwalior_junction"), 20)
    return leg1 + leg2

print("✅ Block 2 Loaded: Transit Matrix & Topological Routing functions active.")
print(f"Sample Transit Check:")
print(f" - Station to Man Mandir: {get_transit_time('gwalior_junction', 'man_mandir_palace')} mins")
print(f" - Fort Plateau to Lashkar (Bada): {get_transit_time('man_mandir_palace', 'maharaj_bada')} mins")
print(f" - Station to Bateshwar (Morena): {get_transit_time('gwalior_junction', 'bateshwar_temples')} mins")
```

```python
# ==========================================
# BLOCK 3: DETERMINISTIC VALIDATION & SAFEGUARDS
# ==========================================

from datetime import datetime
from typing import List, Dict, Tuple, Optional

def parse_time_str(time_str: str) -> int:
    """Converts 'HH:MM' string to minutes from midnight."""
    h, m = map(int, time_str.split(":"))
    return h * 60 + m

def format_mins_to_time(minutes_from_midnight: int) -> str:
    """Converts minutes from midnight back to 'HH:MM AM/PM'."""
    hrs = (minutes_from_midnight // 60) % 24
    mins = minutes_from_midnight % 60
    period = "AM" if hrs < 12 else "PM"
    display_hrs = hrs if hrs <= 12 else hrs - 12
    display_hrs = 12 if display_hrs == 0 else display_hrs
    return f"{display_hrs:02d}:{mins:02d} {period}"

def validate_site_operational_status(
    node_id: str,
    day_of_week: str,
    arrival_time_mins: int,
    stay_duration_mins: int
) -> Tuple[bool, Optional[str]]:
    """
    FR-1: Evaluates deterministic opening rules.
    Returns (is_open, rejection_reason).
    """
    node = GWALIOR_NODES.get(node_id)
    if not node:
        return False, f"Unknown location ID: {node_id}"

    # 1. Day of the week check
    if day_of_week not in node["open_days"]:
        return False, f"CLOSED on {day_of_week}s ({node['jurisdiction']} policy)"

    # 2. Operating hours check
    open_start = parse_time_str(node["open_hours"][0])
    open_end = parse_time_str(node["open_hours"][1])
    departure_time_mins = arrival_time_mins + stay_duration_mins

    if arrival_time_mins < open_start:
        return False, f"Opens at {node['open_hours'][0]}, but arrival scheduled at {format_mins_to_time(arrival_time_mins)}"
    if departure_time_mins > open_end:
        return False, f"Closes at {node['open_hours'][1]}, which overlaps past allowable visit duration"

    return True, None

def check_peripheral_circuit_guardrail(
    candidate_node_ids: List[str],
    time_budget_hours: float
) -> Tuple[bool, Optional[str]]:
    """
    TC-03 Guardrail: Prevents impossible peripheral transit requests (e.g. Bateshwar).
    Bateshwar requires ~150 mins round trip transit + 60 mins visit = 210 mins minimum.
    """
    has_peripheral = any(
        GWALIOR_NODES.get(nid, {}).get("category") == "peripheral_circuit"
        for nid in candidate_node_ids
    )
    if has_peripheral and time_budget_hours < 5.0:
        return False, (
            f"REJECTED: Bateshwar/Mitawali temples sit 35-45 km north in Morena district. "
            f"Round-trip travel requires at least 2.5 hours transit alone. "
            f"Your time budget of {time_budget_hours}h is insufficient. "
            f"City-based heritage alternatives recommended instead."
        )
    return True, None

def filter_accessible_nodes(
    candidate_node_ids: List[str],
    avoid_stairs: bool
) -> Tuple[List[str], List[str]]:
    """
    FR-7 / TC-04: Enforces mobility accessibility constraints.
    Returns (eligible_nodes, excluded_nodes).
    """
    eligible = []
    excluded = []
    for nid in candidate_node_ids:
        node = GWALIOR_NODES.get(nid)
        if avoid_stairs and node and node.get("requires_stairs", False):
            excluded.append(nid)
        else:
            eligible.append(nid)
    return eligible, excluded

print("✅ Block 3 Loaded: Feasibility Validation & Safeguard Guards active.")
```

```python
# ==========================================
# BLOCK 4: DETERMINISTIC ROUTE SEQUENCER & TARIFF ENGINE
# ==========================================

from typing import Dict, List, Any

def plan_margdarshak_itinerary(
    time_budget_hours: float,
    start_time_str: str,
    day_of_week: str,
    origin_node_id: str,
    candidate_node_ids: List[str],
    avoid_stairs: bool = False
) -> Dict[str, Any]:
    """
    Main MargDarshak Optimization Pipeline.
    Enforces FR-1, FR-2, FR-3, FR-4, FR-5, FR-7 deterministically.
    """
    total_budget_mins = int(time_budget_hours * 60)
    current_time_mins = parse_time_str(start_time_str)
    start_time_mins = current_time_mins

    # 1. Peripheral Guardrail Check (TC-03)
    is_safe, guardrail_error = check_peripheral_circuit_guardrail(candidate_node_ids, time_budget_hours)
    if not is_safe:
        return {
            "status": "REJECTED",
            "reason": guardrail_error,
            "itinerary": [],
            "total_duration_mins": 0,
            "total_cost_by_agency": {},
            "embedded_merchants": []
        }

    # 2. Mobility Filtering (FR-7)
    accessible_candidates, excluded_stair_nodes = filter_accessible_nodes(candidate_node_ids, avoid_stairs)

    # 3. Schedule Feasibility Filtering (FR-1)
    valid_candidates = []
    rejected_closed = []

    for nid in accessible_candidates:
        node = GWALIOR_NODES[nid]
        # Tentatively check using an early arrival window
        is_open, close_reason = validate_site_operational_status(
            nid, day_of_week, current_time_mins, node["avg_visit_mins"]
        )
        if is_open:
            valid_candidates.append(nid)
        else:
            rejected_closed.append({"node": node["name"], "reason": close_reason})

    # 4. Deterministic Greedy Route Construction (FR-2, FR-3)
    itinerary = []
    current_loc = origin_node_id
    total_transit_mins = 0
    total_visit_mins = 0
    tariffs = {}
    embedded_merchants = []

    remaining_nodes = valid_candidates.copy()

    while remaining_nodes:
        # Pick the nearest accessible node in transit minutes
        remaining_nodes.sort(key=lambda nid: get_transit_time(current_loc, nid))
        next_node_id = remaining_nodes[0]
        next_node = GWALIOR_NODES[next_node_id]

        transit_time = get_transit_time(current_loc, next_node_id)
        visit_time = next_node["avg_visit_mins"]
        return_to_origin_time = get_transit_time(next_node_id, origin_node_id)

        # Check if adding this node + returning to origin exceeds total budget
        projected_total = (current_time_mins - start_time_mins) + transit_time + visit_time + return_to_origin_time

        if projected_total > total_budget_mins:
            # Cannot fit without exceeding time limit
            remaining_nodes.pop(0)
            continue

        # Re-verify opening hours with exact calculated arrival time
        arrival_mins = current_time_mins + transit_time
        is_still_open, close_err = validate_site_operational_status(
            next_node_id, day_of_week, arrival_mins, visit_time
        )
        if not is_still_open:
            rejected_closed.append({"node": next_node["name"], "reason": close_err})
            remaining_nodes.pop(0)
            continue

        # Add to schedule
        departure_mins = arrival_mins + visit_time
        itinerary.append({
            "stop_number": len(itinerary) + 1,
            "node_id": next_node_id,
            "name": next_node["name"],
            "jurisdiction": next_node["jurisdiction"],
            "transit_mode": "Vehicular/Auto via Urvai Gate" if avoid_stairs and "urvai" in next_node.get("access_point", "") else "Auto-Rickshaw / E-Rickshaw",
            "transit_mins": transit_time,
            "arrival_time": format_mins_to_time(arrival_mins),
            "visit_duration_mins": visit_time,
            "departure_time": format_mins_to_time(departure_mins),
            "entry_fee_inr": next_node["entry_fee_inr"],
            "access_note": "Step-free vehicular access" if avoid_stairs else next_node.get("access_point", "")
        })

        # Tariff by managing body (FR-4)
        agency = next_node["jurisdiction"]
        tariffs[agency] = tariffs.get(agency, 0) + next_node["entry_fee_inr"]

        # Micro-merchants nearby (FR-5)
        for m in next_node.get("merchants", []):
            embedded_merchants.append({
                "merchant_name": m["name"],
                "type": m["type"],
                "near_stop": next_node["name"],
                "walking_distance_m": m["distance_m"]
            })

        current_time_mins = departure_mins
        total_transit_mins += transit_time
        total_visit_mins += visit_time
        current_loc = next_node_id
        remaining_nodes.pop(0)

    # Return transit leg to origin
    final_return_mins = get_transit_time(current_loc, origin_node_id)
    total_transit_mins += final_return_mins
    final_departure_mins = current_time_mins + final_return_mins

    total_elapsed_mins = final_departure_mins - start_time_mins

    return {
        "status": "FEASIBLE" if itinerary else "NO_FEASIBLE_ROUTE",
        "total_budget_mins": total_budget_mins,
        "total_elapsed_mins": total_elapsed_mins,
        "total_visit_mins": total_visit_mins,
        "total_transit_mins": total_transit_mins,
        "return_to_origin_time": format_mins_to_time(final_departure_mins),
        "itinerary": itinerary,
        "total_cost_by_agency": tariffs,
        "total_cost_sum_inr": sum(tariffs.values()),
        "embedded_merchants": embedded_merchants,
        "rejected_closures": rejected_closed,
        "excluded_stairs": excluded_stair_nodes
    }

print("✅ Block 4 Loaded: Route Optimizer, Tariff Engine & Merchant Integrator ready.")
```

```python
# ==============================================================================
# BLOCK 6: INTERACTIVE FOLIUM MAP GENERATOR (COLAB COMPATIBLE)
# ==============================================================================

import folium

def generate_margdarshak_map(itinerary_result: dict, output_filename: str = "margdarshak_route.html"):
    """
    Renders an interactive map using Esri WorldStreetMap tiles.
    100% free, zero API key, no Colab 403 blocks.
    """
    if itinerary_result["status"] != "FEASIBLE" or not itinerary_result["itinerary"]:
        print("Cannot generate map for infeasible or empty itinerary.")
        return None

    # Esri World Street Map: Clean, high-contrast, fully open in Colab
    esri_tile_url = "https://server.arcgisonline.com/ArcGIS/rest/services/World_Street_Map/MapServer/tile/{z}/{y}/{x}"
    esri_attr = "Tiles &copy; Esri &mdash; Source: Esri, USGS, DeLorme, NAVTEQ"

    m = folium.Map(
        location=[26.2230, 78.1700],
        zoom_start=13,
        tiles=esri_tile_url,
        attr=esri_attr
    )

    jurisdiction_colors = {
        "ASI Bhopal Circle": "red",
        "MP State Archaeology": "purple",
        "Scindia Devasthan Trust (Private)": "orange",
        "Gwalior Municipal Corp (GSCDCL)": "green",
        "Indian Railways": "blue"
    }

    route_coords = []

    # Plot Origin (Gwalior Junction)
    origin_node = GWALIOR_NODES["gwalior_junction"]
    origin_lat, origin_lon = origin_node["coords"]
    route_coords.append((origin_lat, origin_lon))

    folium.Marker(
        location=[origin_lat, origin_lon],
        popup="<b>Origin: Gwalior Junction Railway Station</b><br>Transit Departure & Return Hub",
        tooltip="Gwalior Junction (Origin)",
        icon=folium.Icon(color="blue", icon="train", prefix="fa")
    ).add_to(m)

    # Plot Itinerary Stops
    for stop in itinerary_result["itinerary"]:
        node = GWALIOR_NODES[stop["node_id"]]
        lat, lon = node["coords"]
        route_coords.append((lat, lon))
        color = jurisdiction_colors.get(stop["jurisdiction"], "gray")

        popup_html = f"""
        <div style="font-family: Arial; width: 220px;">
            <h4 style="margin: 0; color: #1a365d;">{stop['stop_number']}. {stop['name']}</h4>
            <p style="margin: 4px 0;"><b>Authority:</b> {stop['jurisdiction']}</p>
            <p style="margin: 2px 0;"><b>Arrival:</b> {stop['arrival_time']} | <b>Dep:</b> {stop['departure_time']}</p>
            <p style="margin: 2px 0;"><b>Entry Fee:</b> Rs {stop['entry_fee_inr']}</p>
            <p style="margin: 2px 0; font-size: 11px; color: #4a5568;"><i>{stop['access_note']}</i></p>
        </div>
        """

        folium.Marker(
            location=[lat, lon],
            popup=popup_html,
            tooltip=f"Stop {stop['stop_number']}: {stop['name']}",
            icon=folium.Icon(color=color, icon="landmark", prefix="fa")
        ).add_to(m)

    # Return leg to station
    route_coords.append((origin_lat, origin_lon))

    # Plot Polyline for Transit Route
    folium.PolyLine(
        locations=route_coords,
        color="#0055ff",
        weight=4,
        opacity=0.85,
        dash_array="6",
        tooltip="Transit Path"
    ).add_to(m)

    # Plot Embedded Local Micro-Merchants
    for merchant in itinerary_result.get("embedded_merchants", []):
        host_nodes = [v for k, v in GWALIOR_NODES.items() if v["name"] == merchant["near_stop"]]
        if host_nodes:
            host_node = host_nodes[0]
            m_lat = host_node["coords"][0] + 0.0012
            m_lon = host_node["coords"][1] + 0.0012

            m_icon_color = "darkgreen" if merchant["type"] == "Culinary Heritage" else "cadetblue"

            popup_m = f"""
            <div style="font-family: Arial; width: 200px;">
                <h5 style="margin: 0; color: #22543d;">{merchant['merchant_name']}</h5>
                <p style="margin: 3px 0; font-size: 11px;"><b>Category:</b> {merchant['type']}</p>
                <p style="margin: 2px 0; font-size: 11px;">Within {merchant['walking_distance_m']}m of {merchant['near_stop']}</p>
            </div>
            """
            folium.CircleMarker(
                location=[m_lat, m_lon],
                radius=6,
                color=m_icon_color,
                fill=True,
                fill_color=m_icon_color,
                fill_opacity=0.9,
                popup=popup_m,
                tooltip=f"Local Merchant: {merchant['merchant_name']}"
            ).add_to(m)

    m.save(output_filename)
    print(f"✅ Interactive map rendered and saved to: {output_filename}")
    return m

# Render the Test 1 route generated by Block 5
test_map = generate_margdarshak_map(t1_res, "margdarshak_presentation_route.html")
test_map
```
